```python
from langchain_groq import ChatGroq
import os
from langchain.chat_models import init_chat_model

api_key= os.getenv("GROQ_API_KEY")

llm = ChatGroq(model = "llama-3.1-8b-instant")
```

```python
response = llm.invoke("whats 2 + 2")
response.content
```

```python
from IPython.display import display
from ipywidgets import Text, Button, VBox
import IPython

def create_input_widget():
    text_widget = Text(placeholder='Enter your text here')
    button = Button(description='Submit')
    
    def on_button_click(b):
        print(f"You entered: {text_widget.value}")
        # Process the input here
        
    button.on_click(on_button_click)
    display(VBox([text_widget, button]))

create_input_widget()
```

```python
from IPython.display import display
import ipywidgets as widgets

box = widgets.Text(placeholder="Enter Human Message (/exit to quit)")
display(box)

def on_submit(change):
    if change.value == "/exit":
        print("Exiting...")
        return
    response = llm.invoke(change.value)
    print("AI:", response.content)
    box.value = ""  # clear after submit

box.on_submit(on_submit)
```
