```python
from utils import get_groq_llm
from langchain_core.prompts import ChatPromptTemplate
from langgraph.graph import StateGraph, END
from langgraph.checkpoint.memory import MemorySaver
from typing import TypedDict, Dict, Any
```

```python
llm = get_groq_llm()
```

```python
class QueryAnalysisState(TypedDict):
    user_query: str
    title: str | None
    description: str | None
    linkedin_post: str | None
    hashtags: str| None

# ---- Define prompts ----
title_generator_prompt = ChatPromptTemplate.from_template(
    "Write me Youtube Video Title for this {query}."
)
description_prompt = ChatPromptTemplate.from_template(
    "Write me Youtube Video Description for this {query}."
)
linkedin_post_prompt = ChatPromptTemplate.from_template(
    "I made a Youtube Video on {query}. Now write me a LinkedIn post regarding the topic, give some knowledge and facts about the topic then route it to YT Video."
)
hashtag_generator = ChatPromptTemplate.from_template(
    "Generate me relevant hashtags for video related to {query} on Youtube. Provide me 30 hashtags."
)
# ---- Node functions ----
def generate_title(state: QueryAnalysisState):
    response = (title_generator_prompt | llm).invoke({"query": state["user_query"]})
    return {"title": response.content}

def generate_description(state: QueryAnalysisState):
    response = (description_prompt | llm).invoke({"query": state["user_query"]})
    return {"description": response.content}

def linkedin_post(state: QueryAnalysisState):
    response = (linkedin_post_prompt | llm).invoke({"query": state["user_query"]})
    return {"linkedin_post": response.content}

def generate_hashtags(state: QueryAnalysisState):
    response = (hashtag_generator | llm).invoke({"query": state["user_query"]})
    return {"hashtags": response.content}
    
def merge_results(state: QueryAnalysisState):
    return {
        "final_analysis": {
            "title": state["title"],
            "description": state["description"],
            "linkedin_post": state["linkedin_post"],
            "hashtags": state["hashtags"],
        }
    }

# ---- Build graph ----
builder = StateGraph(QueryAnalysisState)

builder.add_node("title", generate_title)
builder.add_node("description", generate_description)
builder.add_node("post", linkedin_post)
builder.add_node("hashtags", generate_hashtags)
builder.add_node("merge", merge_results)

# Parallel edges
builder.add_edge("__start__", "title")
builder.add_edge("__start__", "description")
builder.add_edge("__start__", "post")
builder.add_edge("__start__", "hashtags")

# Merge all to final node
builder.add_edge("title", "merge")
builder.add_edge("description", "merge")
builder.add_edge("post", "merge")
builder.add_edge("hashtags", "merge")
builder.add_edge("merge", END)

graph = builder.compile(checkpointer=MemorySaver())

# ---- Run ----
result = graph.invoke({"user_query": "Show me all active users older than 30 in New York"})
print(result["final_analysis"])
```
